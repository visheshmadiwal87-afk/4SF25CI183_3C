```python
df = df.drop_duplicates()
```

```python
df = df.drop(columns=['Unnamed: 0'], errors='ignore')
```

```python
df['duration_min'] = df['duration_ms'] / 60000
```

```python
df['popularity_category'] = np.where(
      df['popularity'] >= 70,
          'Popular',
              'Less Popular'
)
```

```python
print(df.isnull().sum())
```

```python
print(df.isnull().mean() * 100)
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].mean()
    )
```

```python
print(df.isnull().sum())
```

```python
top_songs = df.nlargest(10, 'popularity')

print(top_songs[['track_name', 'popularity']])
```

```python
print("Average Popularity:", df['popularity'].mean())
```

```python
energetic_songs = df.nlargest(10, 'energy')

print(energetic_songs[['track_name', 'energy']])
```

```python
print("Average Danceability:", df['danceability'].mean())
```

```python
print(df['popularity_category'].value_counts())
```

```python
top_songs.plot(
      x='track_name',
          y='popularity',
              kind='bar',
                  figsize=(10,5)
                  )
plt.title("Top 10 Popular Songs")
plt.show()
```

```python
plt.hist(df['popularity'], bins=20)
plt.title("Song Popularity Distribution")
plt.xlabel("Popularity")
plt.ylabel("Number of Songs")
plt.show()
```

```python
plt.scatter(df['energy'], df['danceability'])
plt.title("Energy vs Danceability")
plt.xlabel("Energy")
plt.ylabel("Danceability")
plt.show()
```

```python
df['popularity_category'].value_counts().plot(
      kind='pie',
          autopct='%1.1f%%'
          )
plt.title("Song Popularity Categories")
plt.ylabel("")
plt.show()
```
